# Fundamental principle

We cannot randomly change data if we try to achieve a predictable result. We need to identify the parameters of a given data set in terms of statistical properties and have to monitor how these properties changed after any given alteration. Bottomline: Even if we only want to change a single dimension, we have to verify the integrity of any other given dimension.
This may make data generation costly. Depending on what parameters we want to track in the end, it might be useful to regularize data (see comment about ML practices in google docs)
So the goal is to change parameters in a way so that the other parameters stay invariant within a specified margin of error
Furthermore, we have to think about a general tradeoff between accurate generation of ‘randomness’ or how much actual randomness we want to permit

# Null Ratio

A function that is supposed to manipulate the null ratio on data has to do two things:
1. If the ratio in increased, it has to replace 'fitting' entries to maintain the established parameters of a dataset
2. If the ratio is decreased, it has to replace null entries in a way to maintain column correlation and data distribution of a column
We are mostly concerned with the former implementation in this section, as we can outsource the implementation of the distribution generation

```python
def get_null_ratio(df, column):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    if df.empty:
        raise ValueError("Cannot calculate NULL ratio of an empty DataFrame")

    return float(df[column].isna().mean())
```

```python
df = pd.read_csv("airlines.csv")

# Set the NULL ratio of the column 'Delay' to 3%
null_ratio = get_null_ratio(df, "Delay")

# Check the resulting NULL ratio
print(null_ratio)
```

# Number of distinct Values and Number of Entries

```python
import pandas as pd

def count_distinct_values(df, column, include_null=False):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    return df[column].nunique(dropna=not include_null)

def count_entries(df):
    return len(df)


# Usage
df = pd.read_csv("airlines.csv")

distinct_count = count_distinct_values(df, "Delay")
num_entries = count_entries(df)

print(f"Number of Entries: {num_entries}")
print(f"Distinct values: {distinct_count}")
```

# Int typed metrics

```python
def get_value_range(df, column):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    if not pd.api.types.is_integer_dtype(df[column].dtype):
        raise TypeError(f"Column '{column}' must contain integer values")

    if df[column].notna().sum() == 0:
        return None, None

    minimum = df[column].min()
    maximum = df[column].max()

    return minimum, maximum

def get_average(df, column):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    if not pd.api.types.is_numeric_dtype(df[column].dtype) or \
       pd.api.types.is_bool_dtype(df[column].dtype):
        raise TypeError(f"Column '{column}' must contain numeric values")

    if df[column].notna().sum() == 0:
        return None

    return df[column].mean()

def get_variance(df, column):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    if not pd.api.types.is_numeric_dtype(df[column].dtype) or \
       pd.api.types.is_bool_dtype(df[column].dtype):
        raise TypeError(f"Column '{column}' must contain numeric values")

    if df[column].notna().sum() == 0:
        return None

    return df[column].var()

# Usage
df = pd.read_csv("airlines.csv")

minimum, maximum = get_value_range(df, "Delay")
average = get_average(df, "Delay")
variance = get_variance(df, "Delay")

print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Range: {maximum - minimum}")
print(f"Average: {average}")
print(f"Variance: {variance}")
```

# Most Common Values

```python
def get_top_mcv(df, column, n=10):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    return df[column].value_counts().head(n)


# Usage
df = pd.read_csv("airlines.csv")
top_mcv = get_top_mcv(df, "Length")

print(top_mcv)
```

# Maximum and average size of data

```python
def get_entry_size_stats(df, column, encoding="utf-8"):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    values = df[column].dropna()

    if values.empty:
        return None, None

    # Calculate the byte size of each entry
    sizes = values.map(
        lambda value: len(str(value).encode(encoding))
    )

    maximum_size = int(sizes.max())
    average_size = float(sizes.mean())

    return maximum_size, average_size


# Usage
# df = pd.read_csv("airlines.csv", dtype="string") - Might be necessary to determine data type like this
df = pd.read_csv("airlines.csv")
max_size, avg_size = get_entry_size_stats(df, "Delay")

print(f"Maximum entry size: {max_size} bytes")
print(f"Average entry size: {avg_size:.2f} bytes")
```

# Corrleation

```python
def get_correlation(df, column1, column2, method="pearson"):
    for column in [column1, column2]:
        if column not in df.columns:
            raise KeyError(f"Column '{column}' not found")

        if not pd.api.types.is_numeric_dtype(df[column].dtype):
            raise TypeError(f"Column '{column}' must be numeric")

    if method not in ["pearson", "spearman", "kendall"]:
        raise ValueError("Unsupported correlation method")

    return df[column1].corr(df[column2], method=method)


# Usage
df = pd.read_csv("airlines.csv")
correlation = get_correlation(df, "AirportFrom", "AirportTo")

print(f"Correlation: {correlation:.4f}")
```

# Fitness of Normal Distribution

```python
from scipy import stats
import pandas as pd
import numpy as np

def test_normality(df, column, alpha=0.05):
    if column not in df.columns:
        raise KeyError(f"Column '{column}' not found")

    if not pd.api.types.is_numeric_dtype(df[column].dtype):
        raise TypeError(f"Column '{column}' must be numeric")

    values = df[column].dropna().to_numpy()

    if len(values) < 8:
        raise ValueError("At least 8 observations are required")

    if not np.isfinite(values).all():
        raise ValueError("Column contains infinite values")

    if np.ptp(values) == 0:
        raise ValueError("Cannot test a constant column")

    result = stats.normaltest(values)

    return {
        "statistic": result.statistic,
        "p_value": result.pvalue,
        "is_normal": result.pvalue >= alpha
    }


# Usage
df = pd.read_csv("airlines.csv")
result = test_normality(df, "AirportFrom")

print(result)
```

# Empirical Metrics
- Kolmogorov Smirnoff Test
Also Possible: Chi Squared Test, Wassterstein or Energy distance and more, depending what else we want

```python

import pandas as pd
from scipy.stats import ks_2samp

def compare_csv_columns(
    file_a,
    file_b,
    column_a,
    column_b=None,
    alpha=0.05
):
    # Load datasets
    df_a = pd.read_csv(file_a)
    df_b = pd.read_csv(file_b)

    if column_b is None:
        column_b = column_a

    # Extract numerical observations and remove missing values
    sample_a = pd.to_numeric(
        df_a[column_a], errors="coerce"
    ).dropna()

    sample_b = pd.to_numeric(
        df_b[column_b], errors="coerce"
    ).dropna()

    if len(sample_a) == 0 or len(sample_b) == 0:
        raise ValueError("Both columns must contain numerical values.")

    # Perform two-sample KS test
    result = ks_2samp(
        sample_a,
        sample_b,
        alternative="two-sided",
        method="auto"
    )

    if result.pvalue < alpha:
        print("Result: Distributions are significantly different.")
    else:
        print("Result: No significant difference detected.")

    return result

compare_csv_columns("airlines.csv", "output.csv", "Delay")
```

# General algorithm
The idea for data generation is to retrieve the metrics of the data first and to provide margins on how far we can deviate from the original distribution during trace-based data generation. Our goal is to generate the data beforehand, so we can test data before a benchmark run, meaning we can create a validation loop where we can prioritize different metrics during each generation step and build progresssively a new data set.
For instance: We might want to increase null ratios to challenge join estimations. Before that, we would maintain the original average value of a column and its variance. ~~We can normalize the whole data set after introdcuing the null values randomly by adding or substracting the difference in estimated value~~. We would need a more sophisticated algortihm to account for updating variance and covariance. There are ways to calulate the update variance and covariance (Welford algorithm?), which we could maybe use to create pairs of data that maintain the generated data during generation. In another step, we can then remove data that have the least amount of impact on these properties to then restore the amount of entries and Distinct Values that were originally present, while respecting the data range
We might have to adjust the exact validation loop dependent on the type of transformation we want to apply

# Helper Function

```python
def collect_dataset_metrics(
    df,
    columns=None,
    correlation_pairs=None,
    mcv_n=10,
    alpha=0.05,
    encoding="utf-8",
    include_normality=True,
    correlation_methods=("pearson", "spearman", "kendall")
):
    """
    Collect all currently implemented dataset metrics.

    Returns:
        dict: Nested dictionary containing dataset,
              column and correlation metrics.
    """

    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame")

    if df.empty:
        raise ValueError("DataFrame is empty")

    if columns is None:
        columns = list(df.columns)

    for column in columns:
        if column not in df.columns:
            raise KeyError(f"Column '{column}' not found")

    if mcv_n < 1:
        raise ValueError("mcv_n must be at least 1")

    if not 0 < alpha < 1:
        raise ValueError("alpha must be between 0 and 1")

    if correlation_pairs is None:
        correlation_pairs = []

    supported_methods = {"pearson", "spearman", "kendall"}

    if not set(correlation_methods).issubset(supported_methods):
        raise ValueError("Unsupported correlation method")

    metrics = {
        "dataset": {
            "row_count": count_entries(df),
            "column_count": len(df.columns),
            "total_null_count": int(df.isna().sum().sum()),
            "overall_null_ratio": float(
                df.isna().to_numpy().mean()
            )
        },
        "columns": {},
        "correlations": {}
    }

    # Collect individual column metrics
    for column in columns:

        series = df[column]

        # Basic metrics
        column_metrics = {
            "dtype": str(series.dtype),
            "null_count": int(series.isna().sum()),
            "null_ratio": float(series.isna().mean()),

            "distinct_values": count_distinct_values(
                df, column
            ),

            "most_common_values": [
                {"value": value, "count": int(count)}
                for value, count in get_top_mcv(
                    df, column, n=mcv_n
                ).items()
            ]
        }

        # Entry size
        max_size, avg_size = get_entry_size_stats(
            df, column, encoding=encoding
        )

        column_metrics["entry_size"] = {
            "maximum_bytes": max_size,
            "average_bytes": avg_size
        }

        # Integer-specific metrics
        if pd.api.types.is_integer_dtype(series.dtype):

            minimum, maximum = get_value_range(df, column)

            column_metrics["range"] = {
                "minimum": minimum,
                "maximum": maximum
            }
        else:
            column_metrics["range"] = None

        # Numeric metrics
        is_numeric = (
            pd.api.types.is_numeric_dtype(series.dtype)
            and not pd.api.types.is_bool_dtype(series.dtype)
        )

        if is_numeric:

            column_metrics["average"] = get_average(
                df, column
            )

            column_metrics["variance"] = get_variance(
                df, column
            )

            # Normality test
            if include_normality:
                try:
                    column_metrics["normality"] = test_normality(
                        df, column, alpha=alpha
                    )
                except (ValueError, TypeError) as error:
                    column_metrics["normality"] = {
                        "tested": False,
                        "reason": str(error)
                    }
            else:
                column_metrics["normality"] = None

        else:
            column_metrics["average"] = None
            column_metrics["variance"] = None
            column_metrics["normality"] = None

        metrics["columns"][column] = column_metrics

    # Collect correlations between explicitly specified pairs
    for column1, column2 in correlation_pairs:

        pair_name = f"{column1}__{column2}"

        if column1 not in df.columns or column2 not in df.columns:
            raise KeyError(
                f"Invalid correlation pair: {column1}, {column2}"
            )

        metrics["correlations"][pair_name] = {}

        for method in correlation_methods:

            correlation = get_correlation(
                df, column1, column2, method=method
            )

            metrics["correlations"][pair_name][method] = (
                float(correlation)
                if pd.notna(correlation)
                else None
            )

    return metrics


from IPython.display import display, Markdown

def display_dataset_metrics(metrics):
    """
    Display collected dataset metrics in neatly formatted tables.

    Args:
        metrics (dict): Output of collect_dataset_metrics()

    Returns:
        dict: Dictionary containing the generated display tables.
    """

    if not isinstance(metrics, dict):
        raise TypeError("Metrics must be a dictionary")

    tables = {}

    # --------------------------------------------------
    # 1. Dataset overview
    # --------------------------------------------------
    display(Markdown("## Dataset Overview"))

    dataset = metrics.get("dataset", {})
    tables["dataset"] = pd.DataFrame.from_dict(
        dataset,
        orient="index",
        columns=["Value"]
    )

    display(tables["dataset"])

    # --------------------------------------------------
    # 2. Individual column metrics
    # --------------------------------------------------
    display(Markdown("## Column Metrics"))

    column_rows = {}

    for column, values in metrics.get("columns", {}).items():

        row = {
            "Data type": values.get("dtype"),
            "NULL count": values.get("null_count"),
            "NULL ratio": values.get("null_ratio"),
            "Distinct values": values.get("distinct_values"),
            "Average": values.get("average"),
            "Variance": values.get("variance")
        }

        # Range
        value_range = values.get("range")
        if isinstance(value_range, dict):
            row["Minimum"] = value_range.get("minimum")
            row["Maximum"] = value_range.get("maximum")

        # Entry size
        entry_size = values.get("entry_size")
        if isinstance(entry_size, dict):
            row["Max size (bytes)"] = entry_size.get("maximum_bytes")
            row["Avg size (bytes)"] = entry_size.get("average_bytes")

        # Most common values
        mcv = values.get("most_common_values", [])
        row["Most common values"] = ", ".join(
            f"{item['value']} (n={item['count']})"
            for item in mcv
        )

        # Normality
        normality = values.get("normality")

        if isinstance(normality, dict):
            row["Normality statistic"] = normality.get("statistic")
            row["Normality p-value"] = normality.get("p_value")
            row["Normally distributed"] = normality.get("is_normal")

        column_rows[column] = row

    tables["columns"] = pd.DataFrame.from_dict(
        column_rows,
        orient="index"
    )

    display(tables["columns"])

    # --------------------------------------------------
    # 3. Correlations
    # --------------------------------------------------
    display(Markdown("## Correlations"))

    correlation_rows = []

    for pair, methods in metrics.get("correlations", {}).items():

        row = {"Column pair": pair}
        row.update(methods)
        correlation_rows.append(row)

    tables["correlations"] = pd.DataFrame(
        correlation_rows
    )

    display(tables["correlations"])

    return tables

def load_input_output(input_file, output_file, column):
    input_path = Path(input_file)
    output_path = Path(output_file)

    if not input_path.is_file():
        raise FileNotFoundError(f"Input file not found: {input_file}")

    if input_path.resolve() == output_path.resolve():
        raise ValueError("Input and output files must be different")

    input_df = pd.read_csv(input_path)

    if column not in input_df.columns:
        raise KeyError(f"Column '{column}' not found in input file")

    output_path.parent.mkdir(parents=True, exist_ok=True)

    if not output_path.exists():
        shutil.copy2(input_path, output_path)

    output_df = pd.read_csv(output_path)

    if column not in output_df.columns:
        raise KeyError(f"Column '{column}' not found in output file")

    if len(input_df) != len(output_df):
        raise ValueError("Input and output must have the same number of rows")

    return input_df, output_df, output_path


def format_ks_result(
    result,
    alpha=0.05,
    sample_a_name="Sample A",
    sample_b_name="Sample B"
):
    """
    Formats the result of scipy.stats.ks_2samp()
    into a readable command-line report.

    Returns:
        str: Formatted test report.
    """

    statistic = float(result.statistic)
    p_value = float(result.pvalue)

    if not 0 < alpha < 1:
        raise ValueError("Alpha must be between 0 and 1")

    if not (0 <= statistic <= 1 and 0 <= p_value <= 1):
        raise ValueError("Invalid KS test result")

    # Determine statistical conclusion
    if p_value < alpha:
        conclusion = "REJECT H0"
        interpretation = (
            "Statistically significant evidence that the "
            "distributions are different."
        )
    else:
        conclusion = "CANNOT REJECT H0"
        interpretation = (
            "No statistically significant evidence of "
            "a difference between the distributions."
        )

    # Format report
    separator = "=" * 65

    report = [
        separator,
        "          KOLMOGOROV-SMIRNOV TEST REPORT",
        separator,
        "",
        f"Comparing: {sample_a_name} vs. {sample_b_name}",
        "",
        "-" * 65,
        "TEST METRICS",
        "-" * 65,
        f"KS Statistic (D): {statistic:.6f}",
        f"P-value:          {p_value:.6e}",
        f"Significance:     {alpha:.4f}",
        "",
        "-" * 65,
        "INTERPRETATION",
        "-" * 65,
        "",
        "1. KS Statistic (D):",
        "   Maximum distance between the two empirical",
        "   cumulative distribution functions.",
        "   Range: 0 (identical ECDFs) to 1 (maximum separation).",
        "",
        "2. P-value:",
        "   Measures how extreme the observed difference would be",
        "   under the assumption that both samples come from",
        "   the same underlying distribution.",
        "",
        "3. Significance level (alpha):",
        "   Threshold used to decide whether to reject H0.",
        "",
        "-" * 65,
        "TEST RESULT",
        "-" * 65,
        f"Decision: {conclusion}",
        "",
        interpretation,
        "",
        separator
    ]

    # Include location of maximum difference if available
    location = getattr(result, "statistic_location", None)

    if location is not None:
        report.insert(
            10,
            f"Maximum distance occurs at: {location}"
        )

    sign = getattr(result, "statistic_sign", None)

    if sign is not None:
        direction = (
            f"{sample_a_name} ECDF is above {sample_b_name} ECDF."
            if sign == 1
            else f"{sample_b_name} ECDF is above {sample_a_name} ECDF."
        )

        report.insert(11, f"Direction: {direction}")

    return "\n".join(report)

def generate_bounded_noise(
    n,
    reference_variance,
    variance_tolerance=0.05,
    absolute_variance_tolerance=0.0,
    randomness=1.0,
    seed=43,
    rng=None
):
    """
    Generates bounded random perturbations with zero mean.

    The maximum perturbation is derived from the
    permitted variance deviation.

    randomness:
        0.0 = no randomness
        1.0 = maximum permitted randomness
    """

    if n < 2:
        raise ValueError(
            "At least two observations are required"
        )

    if (
        reference_variance < 0
        or not np.isfinite(reference_variance)
    ):
        raise ValueError(
            "Reference variance must be finite and non-negative"
        )

    if (
        variance_tolerance < 0
        or absolute_variance_tolerance < 0
        or not np.isfinite(variance_tolerance)
        or not np.isfinite(absolute_variance_tolerance)
    ):
        raise ValueError(
            "Variance tolerances must be finite and non-negative"
        )

    if not 0 <= randomness <= 1:
        raise ValueError(
            "Randomness must be between 0 and 1"
        )

    if rng is None:
        rng = np.random.default_rng(seed)

    # Calculate allowed variance deviation
    allowed_variance_error = (
        reference_variance * variance_tolerance
        + absolute_variance_tolerance
    )

    reference_std = np.sqrt(reference_variance)

    # Maximum standard deviation of random noise
    max_noise_std = (
        np.sqrt(reference_variance + allowed_variance_error)
        - reference_std
    )

    # Maximum individual perturbation
    max_deviation = (
        max_noise_std * np.sqrt((n - 1) / n)
    )

    if max_deviation == 0 or randomness == 0:
        return np.zeros(n), max_deviation

    # Generate random values in opposite pairs
    # This guarantees that the perturbations sum to zero
    half = rng.uniform(-1, 1, size=n // 2)

    noise = np.concatenate([half, -half])

    if n % 2:
        noise = np.append(noise, 0.0)

    rng.shuffle(noise)

    peak = np.max(np.abs(noise))

    if peak == 0:
        return np.zeros(n), max_deviation

    # Determine actual perturbation strength
    actual_deviation = (
        rng.uniform(0, randomness) * max_deviation
    )

    noise = noise / peak * actual_deviation

    return noise, max_deviation


def get_valid_values(input_df, output_df, column):

    if (
        not pd.api.types.is_numeric_dtype(input_df[column])
        or pd.api.types.is_bool_dtype(input_df[column])
    ):
        raise TypeError("Input column must be numeric")

    if (
        not pd.api.types.is_numeric_dtype(output_df[column])
        or pd.api.types.is_bool_dtype(output_df[column])
    ):
        raise TypeError("Output column must be numeric")

    # Only consider rows that are valid in BOTH datasets
    valid_mask = (
        input_df[column].notna()
        & output_df[column].notna()
    )

    # Extract corresponding values
    input_values = input_df.loc[
        valid_mask, column
    ].to_numpy(dtype=float)

    output_values = output_df.loc[
        valid_mask, column
    ].to_numpy(dtype=float)

    if len(input_values) < 2:
        raise ValueError("At least two valid observations required")

    if not np.isfinite(input_values).all():
        raise ValueError("Input contains infinite values")

    if not np.isfinite(output_values).all():
        raise ValueError("Output contains infinite values")

    return valid_mask, input_values, output_values
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

def set_null_ratio(
    input_file,
    output_file,
    column,
    target_ratio,
    seed=43
):
    """
    Modifies the NULL ratio in an output CSV based exclusively
    on the original input dataset.

    The input file is the reference.
    The output file is the modification target.
    """

    input_path = Path(input_file)
    output_path = Path(output_file)

    if not input_path.is_file():
        raise FileNotFoundError(f"Input file not found: {input_file}")

    if input_path.resolve() == output_path.resolve():
        raise ValueError("Input and output files must be different")

    if not 0 <= target_ratio <= 1:
        raise ValueError("Target ratio must be between 0 and 1")

    # Load original reference dataset
    input_df = pd.read_csv(input_path)

    if column not in input_df.columns:
        raise KeyError(f"Column '{column}' not found")

    # Create output file if it does not exist
    output_path.parent.mkdir(parents=True, exist_ok=True)

    if not output_path.exists():
        shutil.copy2(input_path, output_path)

    # Load dataset to be modified
    output_df = pd.read_csv(output_path)

    if column not in output_df.columns:
        raise KeyError(f"Column '{column}' not found in output")

    if len(input_df) != len(output_df):
        raise ValueError("Input and output must have the same number of rows")

    rng = np.random.default_rng(seed)

    # ------------------------------------------
    # DERIVE STATISTICS FROM INPUT ONLY
    # ------------------------------------------

    current_ratio = get_null_ratio(input_df, column)

    current_nulls = int(input_df[column].isna().sum())
    target_nulls = round(len(input_df) * target_ratio)

    col_idx = output_df.columns.get_loc(column)

    # ------------------------------------------
    # INCREASE NULL RATIO
    # ------------------------------------------

    if target_nulls > current_nulls:

        series = output_df[column]

        if pd.api.types.is_unsigned_integer_dtype(series.dtype):
            output_df[column] = series.astype("UInt64")

        elif pd.api.types.is_integer_dtype(series.dtype):
            output_df[column] = series.astype("Int64")

        elif pd.api.types.is_bool_dtype(series.dtype):
            output_df[column] = series.astype("boolean")

        # Candidate positions derived from INPUT
        candidates = np.flatnonzero(
            input_df[column].notna().to_numpy()
        )

        positions = rng.choice(
            candidates,
            size=target_nulls - current_nulls,
            replace=False
        )

        # Apply changes to OUTPUT
        output_df.iloc[positions, col_idx] = pd.NA

        action = "increased"

    # ------------------------------------------
    # DECREASE NULL RATIO
    # ------------------------------------------

    elif target_nulls < current_nulls:

        if (
            not pd.api.types.is_numeric_dtype(input_df[column])
            or pd.api.types.is_bool_dtype(input_df[column])
            or pd.api.types.is_complex_dtype(input_df[column])
        ):
            raise TypeError(
                "Decreasing NULL ratio requires a real numeric column"
            )

        # Determine range from INPUT
        source_values = input_df[column].dropna()

        if source_values.empty:
            raise ValueError("Input column contains no valid values")

        minimum = source_values.min()
        maximum = source_values.max()

        if not np.isfinite([minimum, maximum]).all():
            raise ValueError("Input column contains infinite values")

        # Candidate positions derived from INPUT
        candidates = np.flatnonzero(
            input_df[column].isna().to_numpy()
        )

        number_to_fill = current_nulls - target_nulls

        positions = rng.choice(
            candidates,
            size=number_to_fill,
            replace=False
        )

        # Generate values within original input range
        integral_values = np.all(
            np.equal(source_values.to_numpy() % 1, 0)
        )

        if integral_values:
            replacements = rng.integers(
                int(minimum),
                int(maximum),
                size=number_to_fill,
                endpoint=True
            )
        else:
            replacements = rng.uniform(
                float(minimum),
                float(maximum),
                size=number_to_fill
            )

        # Apply changes to OUTPUT
        output_df.iloc[positions, col_idx] = replacements

        action = "decreased"

    else:
        action = "unchanged"

    # Save modified output
    output_df.to_csv(output_path, index=False)

    return {
        "input_file": str(input_path),
        "output_file": str(output_path),
        "column": column,
        "original_null_ratio": current_ratio,
        "target_null_ratio": target_ratio,
        "output_null_ratio": get_null_ratio(output_df, column),
        "action": action
    }

result = set_null_ratio(
    input_file="airlines.csv",
    output_file="output.csv",
    column="Delay",
    target_ratio=0.10
)

print(result)
```

```python
def set_average(
    input_file,
    output_file,
    column,
    target_average,
    mean_tolerance=0.0,
    variance_tolerance=0.05,
    absolute_variance_tolerance=0.0,
    randomness=1.0,
    seed=43
):
    """
    Changes the average of a column.

    Reference statistics are derived from the input file.
    Modifications are applied to the output file.

    A constant shift changes the average, while bounded
    random perturbations introduce controlled variety.
    """

    if not np.isfinite(target_average):
        raise ValueError(
            "Target average must be finite"
        )

    if (
        mean_tolerance < 0
        or not np.isfinite(mean_tolerance)
    ):
        raise ValueError(
            "Mean tolerance must be finite and non-negative"
        )

    if not 0 <= randomness <= 1:
        raise ValueError(
            "Randomness must be between 0 and 1"
        )

    input_df, output_df, output_path = load_input_output(
        input_file, output_file, column
    )

    # Extract valid observations
    valid_mask, source, output_values = get_valid_values(
        input_df, output_df, column
    )

    # Calculate original statistics from INPUT
    original_average = np.mean(source)
    original_variance = np.var(source, ddof=1)

    rng = np.random.default_rng(seed)

    # Select an actual target average within tolerance
    actual_target = target_average + (
        rng.uniform(-mean_tolerance, mean_tolerance)
        * randomness
    )

    # Calculate required average shift
    shift = actual_target - original_average

    # Generate bounded random perturbations
    noise, max_deviation = generate_bounded_noise(
        n=len(source),
        reference_variance=original_variance,
        variance_tolerance=variance_tolerance,
        absolute_variance_tolerance=absolute_variance_tolerance,
        randomness=randomness,
        rng=rng
    )

    # Apply shift and perturbations
    modified_values = output_values + shift + noise

    # Only modify valid output positions
    result = output_df[column].astype(float).copy()

    result.loc[valid_mask] = modified_values

    output_df[column] = result

    # Save output
    output_df.to_csv(output_path, index=False)

    return {
        "column": column,
        "original_average": original_average,
        "target_average": target_average,
        "actual_target_average": actual_target,
        "applied_shift": shift,
        "original_variance": original_variance,
        "max_random_deviation": max_deviation,
        "output_average": get_average(output_df, column),
        "output_file": str(output_path)
    }

def set_variance(
    input_file,
    output_file,
    column,
    target_variance,
    target_average=None,
    mean_tolerance=0.0,
    variance_tolerance=0.05,
    absolute_variance_tolerance=0.0,
    randomness=1.0,
    seed=43
):
    """
    Changes the variance of a column.

    Reference statistics are derived from the input file.
    Modifications are applied to the output file.

    Variance is changed by scaling deviations around
    the original average.

    Bounded random perturbations introduce variety.
    """

    if (
        not np.isfinite(target_variance)
        or target_variance < 0
    ):
        raise ValueError(
            "Target variance must be finite and non-negative"
        )

    if (
        mean_tolerance < 0
        or not np.isfinite(mean_tolerance)
    ):
        raise ValueError(
            "Mean tolerance must be finite and non-negative"
        )

    if not 0 <= randomness <= 1:
        raise ValueError(
            "Randomness must be between 0 and 1"
        )

    input_df, output_df, output_path = load_input_output(
        input_file, output_file, column
    )

    # Extract valid observations
    valid_mask, source, output_values = get_valid_values(
        input_df, output_df, column
    )

    # Calculate original statistics from INPUT
    original_average = np.mean(source)
    original_variance = np.var(source, ddof=1)

    # Handle constant columns
    if original_variance == 0 and target_variance > 0:
        raise ValueError(
            "Cannot increase variance from a constant column "
            "using scaling alone"
        )

    rng = np.random.default_rng(seed)

    # Select target average
    if target_average is None:
        target_average = original_average

    if not np.isfinite(target_average):
        raise ValueError(
            "Target average must be finite"
        )

    actual_average = target_average + (
        rng.uniform(-mean_tolerance, mean_tolerance)
        * randomness
    )

    # Calculate scaling factor
    if original_variance == 0:
        scaling_factor = 0.0
    else:
        scaling_factor = np.sqrt(
            target_variance / original_variance
        )

    # Scale deviations around original average
    scaled_values = (
        actual_average
        + (output_values - original_average) * scaling_factor
    )

    # Generate bounded random perturbations
    noise, max_deviation = generate_bounded_noise(
        n=len(source),
        reference_variance=target_variance,
        variance_tolerance=variance_tolerance,
        absolute_variance_tolerance=absolute_variance_tolerance,
        randomness=randomness,
        rng=rng
    )

    # Apply perturbations
    modified_values = scaled_values + noise

    # Only modify valid output positions
    result = output_df[column].astype(float).copy()

    result.loc[valid_mask] = modified_values

    output_df[column] = result

    # Save output
    output_df.to_csv(output_path, index=False)

    return {
        "column": column,
        "original_average": original_average,
        "original_variance": original_variance,
        "target_average": actual_average,
        "target_variance": target_variance,
        "scaling_factor": scaling_factor,
        "max_random_deviation": max_deviation,
        "output_variance": get_variance(output_df, column),
        "output_file": str(output_path)
    }


input_file = "airlines.csv"
output_file = "output.csv"

# First change variance
variance_result = set_variance(
    input_file,
    output_file,
    column="Delay",
    target_variance=100.0
)

# Then change average
average_result = set_average(
    input_file,
    output_file,
    column="Delay",
    target_average=20.0
)

print(variance_result)
print(average_result)
```
